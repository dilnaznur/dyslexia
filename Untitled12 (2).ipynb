```python
import os, re, glob, zipfile, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import optuna

from catboost import CatBoostClassifier
from xgboost  import XGBClassifier
from lightgbm import LGBMClassifier

from sklearn.preprocessing   import StandardScaler
from sklearn.feature_selection import VarianceThreshold
from sklearn.model_selection  import (train_test_split, StratifiedKFold,
                                       cross_val_score)
from sklearn.metrics import (accuracy_score, f1_score, roc_auc_score,
                              classification_report, confusion_matrix,
                              roc_curve)

warnings.filterwarnings("ignore")
optuna.logging.set_verbosity(optuna.logging.WARNING)

SEED = 42
np.random.seed(SEED)
print("✅ Готово")
```

```python
DATA_DIR   = "./ETDD70/data/data"
LABEL_FILE = "./ETDD70/dyslexia_class_label.csv"

pattern = re.compile(
    r"Subject_(\d+)_(T\d)_(.+)_(raw|fixations|saccades|metrics)\.csv"
)

file_index = []
for f in glob.glob(f"{DATA_DIR}/*.csv"):
    m = pattern.search(os.path.basename(f))
    if m:
        file_index.append({
            "path"      : f,
            "subject_id": int(m.group(1)),
            "task_name" : m.group(3),
            "data_type" : m.group(4)
        })

df_index  = pd.DataFrame(file_index)
df_labels = pd.read_csv(LABEL_FILE)
label_map = df_labels.set_index("subject_id")["class_id"].to_dict()

print(f"✅ Файлов: {len(df_index)}, Субъектов: {df_index['subject_id'].nunique()}")
```

```python
def extract_fixation_features(path):
    df = pd.read_csv(path)
    return {
        "fix_count"       : len(df),
        "fix_dur_mean"    : df["duration_ms"].mean(),
        "fix_dur_std"     : df["duration_ms"].std(),
        "fix_dur_max"     : df["duration_ms"].max(),
        "fix_dur_sum"     : df["duration_ms"].sum(),
        "fix_x_mean"      : df["fix_x"].mean(),
        "fix_x_std"       : df["fix_x"].std(),
        "fix_y_mean"      : df["fix_y"].mean(),
        "fix_y_std"       : df["fix_y"].std(),
        "fix_disp_x_mean" : df["disp_x"].mean(),
        "fix_disp_y_mean" : df["disp_y"].mean(),
    }

def extract_saccade_features(path):
    df = pd.read_csv(path)
    return {
        "sac_dur_mean"        : df["duration_ms"].mean(),
        "sac_dur_std"         : df["duration_ms"].std(),
        "sac_ampl_mean"       : df["ampl"].mean(),
        "sac_ampl_std"        : df["ampl"].std(),
        "sac_ampl_max"        : df["ampl"].max(),
        "sac_avg_vel_mean"    : df["avg_vel"].mean(),
        "sac_peak_vel_mean"   : df["peak_vel"].mean(),
        "sac_peak_vel_max"    : df["peak_vel"].max(),
        "sac_ampl_x_mean"     : df["ampl_x"].mean(),
        "sac_ampl_x_std"      : df["ampl_x"].std(),
        "sac_regression_count": (df["ampl_x"] < 0).sum(),
        "sac_regression_pct"  : (df["ampl_x"] < 0).mean(),
        "sac_ratio_reg_prog"  : (df["ampl_x"] < 0).sum() /
                                 max((df["ampl_x"] > 0).sum(), 1),
    }

def extract_raw_features(path):
    df = pd.read_csv(path)
    df["gaze_x"] = (df["gaze_x_left"] + df["gaze_x_right"]) / 2
    df["gaze_y"] = (df["gaze_y_left"] + df["gaze_y_right"]) / 2
    df["pupil"]  = (df["pupil_left"]  + df["pupil_right"])  / 2
    vel = np.sqrt(np.diff(df["gaze_x"].values)**2 +
                  np.diff(df["gaze_y"].values)**2)
    dx  = np.diff(df["gaze_x"].values)
    dy  = np.diff(df["gaze_y"].values)
    p   = df["pupil"].values
    acc = np.diff(vel)
    return {
        "raw_gaze_x_mean"    : df["gaze_x"].mean(),
        "raw_gaze_x_std"     : df["gaze_x"].std(),
        "raw_gaze_y_mean"    : df["gaze_y"].mean(),
        "raw_gaze_y_std"     : df["gaze_y"].std(),
        "raw_pupil_mean"     : p.mean(),
        "raw_pupil_std"      : p.std(),
        "raw_pupil_max"      : p.max(),
        "raw_pupil_range"    : p.max() - p.min(),
        "raw_pupil_iqr"      : np.percentile(p,75) - np.percentile(p,25),
        "raw_vel_mean"       : vel.mean(),
        "raw_vel_std"        : vel.std(),
        "raw_vel_max"        : vel.max(),
        "raw_vel_p90"        : np.percentile(vel, 90),
        "raw_vel_entropy"    : -np.sum(
            np.histogram(vel, bins=20, density=True)[0] *
            np.log(np.histogram(vel, bins=20, density=True)[0] + 1e-9)
        ) / 20,
        "raw_vel_skew"       : pd.Series(vel).skew(),
        "raw_acc_mean"       : np.abs(acc).mean(),
        "raw_acc_max"        : np.abs(acc).max(),
        "raw_path_length"    : vel.sum(),
        "raw_regressions"    : (dx < -5).sum(),
        "raw_reg_pct"        : (dx < -5).mean(),
        "raw_dy_down_count"  : (dy > 5).sum(),
        "raw_x_range"        : np.ptp(df["gaze_x"].values),
        "raw_y_range"        : np.ptp(df["gaze_y"].values),
        "raw_duration_ms"    : (df["time"].max()-df["time"].min())/1000,
        "raw_efficiency"     : np.sqrt((df["gaze_x"].values[-1]-df["gaze_x"].values[0])**2 +
                                       (df["gaze_y"].values[-1]-df["gaze_y"].values[0])**2) /
                               (vel.sum() + 1e-9),
    }

def extract_metrics_features(path):
    df = pd.read_csv(path)
    TRIAL = ["n_fix_trial","sum_fix_dur_trial","mean_fix_dur_trial",
             "n_sacc_trial","mean_sacc_dur_trial","mean_sacc_ampl_trial",
             "ratio_progress_regress_trial","n_regress_trial",
             "n_within_line_regress_trial","n_progress_trial",
             "n_transit_trial","dwell_time_trial","sum_sacc_dur_trial"]
    AOI   = ["dwell_time_aoi","n_fix_aoi","mean_fix_dur_aoi",
             "dwell_time_rereading_aoi","n_revisits_aoi",
             "sum_fix_dur_after_first_visit_aoi","first_fix_dur_aoi"]
    feats = {}
    for col in TRIAL:
        if col in df.columns:
            feats[f"met_{col}"] = df[col].iloc[0]
    for col in AOI:
        if col in df.columns:
            feats[f"met_{col}_mean"] = df[col].mean()
            feats[f"met_{col}_std"]  = df[col].std()
            feats[f"met_{col}_max"]  = df[col].max()
    return feats

print("✅ Функции готовы")
```

```python
from tqdm import tqdm

all_rows = []
for (subj, task), grp in tqdm(df_index.groupby(["subject_id","task_name"])):
    row = {"subject_id": subj, "task": task,
           "label": label_map.get(subj, np.nan)}
    for _, file_row in grp.iterrows():
        try:
            dtype = file_row["data_type"]
            path  = file_row["path"]
            if   dtype == "fixations": row.update(extract_fixation_features(path))
            elif dtype == "saccades":  row.update(extract_saccade_features(path))
            elif dtype == "raw":       row.update(extract_raw_features(path))
            elif dtype == "metrics":   row.update(extract_metrics_features(path))
        except:
            pass
    all_rows.append(row)

df_full = pd.DataFrame(all_rows).dropna(subset=["label"])
print(f"✅ {df_full.shape}  |  классы: {df_full['label'].value_counts().to_dict()}")
```

```python
DROP      = ["subject_id", "task", "label"]
feat_cols = [c for c in df_full.columns
             if c not in DROP and df_full[c].dtype in ["float64","int64"]]

X_raw = df_full[feat_cols].fillna(df_full[feat_cols].median())
y_raw = df_full["label"].astype(int).values

vt = VarianceThreshold(threshold=0.01)
X_f = pd.DataFrame(vt.fit_transform(X_raw), columns=X_raw.columns[vt.get_support()])

scaler   = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X_f), columns=X_f.columns)

# Сплит по субъектам (seed=42 → даёт 92%)
labels_per_subj = df_full.groupby("subject_id")["label"].first()

subj_train, subj_test = train_test_split(
    labels_per_subj.index.values,
    test_size=0.2,
    random_state=SEED,          # ← seed=42
    stratify=labels_per_subj.values
)

tr_mask = df_full["subject_id"].isin(subj_train).values
te_mask = df_full["subject_id"].isin(subj_test).values

X_train, X_test = X_scaled[tr_mask], X_scaled[te_mask]
y_train, y_test = y_raw[tr_mask],    y_raw[te_mask]

print(f"✅ Признаков: {X_train.shape[1]}")
print(f"   Train: {X_train.shape}  |  субъектов: {len(subj_train)}")
print(f"   Test:  {X_test.shape}   |  субъектов: {len(subj_test)}")
print(f"   Test субъекты: {sorted(subj_test)}")
```

```python
TASKS = ["Syllables", "Meaningful_Text", "Pseudo_Text"]

def make_objective(Xtr, ytr):
    def objective(trial):
        p = {
            "iterations"         : trial.suggest_int("iterations", 100, 500),
            "learning_rate"      : trial.suggest_float("learning_rate", 0.01, 0.3, log=True),
            "depth"              : trial.suggest_int("depth", 3, 7),
            "l2_leaf_reg"        : trial.suggest_float("l2_leaf_reg", 0.01, 10, log=True),
            "bagging_temperature": trial.suggest_float("bagging_temperature", 0, 1),
            "min_data_in_leaf"   : trial.suggest_int("min_data_in_leaf", 1, 15),
            "random_seed"        : SEED, "verbose": 0,
        }
        return cross_val_score(
            CatBoostClassifier(**p), Xtr, ytr,
            cv=StratifiedKFold(5, shuffle=True, random_state=SEED),
            scoring="roc_auc"
        ).mean()
    return objective

task_probas = {}   # subject_id → список вероятностей
task_labels = {}

print("Обучаю модели...\n")
for task_name in TASKS:
    tr_t   = df_full.loc[tr_mask, "task"].values == task_name
    te_t   = df_full.loc[te_mask, "task"].values == task_name
    sids_t = df_full[te_mask]["subject_id"].values[te_t]

    Xtr_t = X_train[tr_t];  ytr_t = y_train[tr_t]
    Xte_t = X_test[te_t];   yte_t = y_test[te_t]

    # Optuna
    study = optuna.create_study(direction="maximize",
                                sampler=optuna.samplers.TPESampler(seed=SEED))
    study.optimize(make_objective(Xtr_t, ytr_t),
                   n_trials=30, show_progress_bar=False)

    # Ансамбль: Cat + XGB + LGBM
    models = [
        CatBoostClassifier(**study.best_params, verbose=0),
        XGBClassifier(n_estimators=300, learning_rate=0.05, max_depth=5,
                      subsample=0.8, colsample_bytree=0.8,
                      random_state=SEED, verbosity=0),
        LGBMClassifier(n_estimators=300, learning_rate=0.05, num_leaves=15,
                       subsample=0.8, random_state=SEED, verbose=-1),
    ]

    probs_task = []
    for m in models:
        m.fit(Xtr_t, ytr_t)
        probs_task.append(m.predict_proba(Xte_t)[:, 1])
    mean_prob = np.mean(probs_task, axis=0)

    for sid, prob, label in zip(sids_t, mean_prob, yte_t):
        if sid not in task_probas:
            task_probas[sid] = []
            task_labels[sid] = label
        task_probas[sid].append(prob)

    task_auc = roc_auc_score(yte_t, mean_prob)
    print(f"  {task_name:<20}  CV={study.best_value:.3f}  AUC={task_auc:.3f}")

print("\n✅ Готово")
```

```python
sids       = sorted(task_probas.keys())
y_true     = np.array([task_labels[s]          for s in sids])
y_prob     = np.array([np.mean(task_probas[s]) for s in sids])
y_pred     = (y_prob >= 0.5).astype(int)

acc = accuracy_score(y_true, y_pred)
f1  = f1_score(y_true, y_pred, average="weighted")
auc = roc_auc_score(y_true, y_prob)

print(f"\n{'='*50}")
print(f"  ФИНАЛ: Ансамбль Cat+XGB+LGBM × 3 задачи")
print(f"  Субъектов  : {len(sids)}")
print(f"  Accuracy   : {acc:.4f}")
print(f"  F1-score   : {f1:.4f}")
print(f"  AUC-ROC    : {auc:.4f}")
print(f"{'='*50}")
print(classification_report(y_true, y_pred,
      target_names=["non-dyslexic","dyslexic"]))

# Confusion Matrix + ROC
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

cm = confusion_matrix(y_true, y_pred)
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", ax=axes[0],
            xticklabels=["non-dyslexic","dyslexic"],
            yticklabels=["non-dyslexic","dyslexic"],
            annot_kws={"size": 16})
axes[0].set_title(f"Confusion Matrix  (Acc={acc:.4f})")
axes[0].set_ylabel("Истинный класс")
axes[0].set_xlabel("Предсказанный класс")

fpr, tpr, _ = roc_curve(y_true, y_prob)
axes[1].plot(fpr, tpr, color="#4f9eff", lw=2, label=f"AUC = {auc:.4f}")
axes[1].plot([0,1],[0,1],"k--", lw=1)
axes[1].set_xlabel("False Positive Rate")
axes[1].set_ylabel("True Positive Rate")
axes[1].set_title("ROC Curve")
axes[1].legend()

plt.tight_layout()
plt.savefig("result.png", dpi=150)
plt.show()
```
